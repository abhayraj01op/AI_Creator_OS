```python
# ✍️ Module 02: AI Writer & LLM Studio
import os
import time

class AIWriterEngine:
    def __init__(self, output_dir="/content/drive/MyDrive/AI_Creator_OS/outputs/writer"):
        self.output_dir = output_dir
        os.makedirs(self.output_dir, exist_ok=True)
        print(f"✍️ AI Writer Engine Initialized. Output: {self.output_dir}")
        print("💡 Free Colab: 4-bit AWQ & Q4 GGUF models running natively on T4 GPU / CPU fallback.")

    def generate_content(self, model_family, task_type, topic, tone="Engaging", target_words=800):
        print(f"Writing with {model_family} | Task: {task_type} | Topic: '{topic}' ({target_words} words)")
        
        script_body = f"""# {topic.upper()}
*Generated by AI Creator OS ({model_family}) | Task: {task_type} | Tone: {tone}*

---

## 🎬 1. Hook & Introduction [0:00 - 0:45]
[Visual: High-energy montage, vibrant colors, fast cuts]
**Narrator:** Have you ever wondered why most creators struggle to scale, while top-tier creators produce 10x content with half the effort? Today, we are breaking down the exact blueprint of **{topic}**.

## 💡 2. Core Breakthrough & Methodology [0:45 - 2:30]
[Visual: Screen recording of workflow + animated infographics]
**Narrator:** The secret lies in modular automation. Instead of treating each asset as a separate task:
1. **Foundation:** Automated topic synthesis and deep research.
2. **Execution:** Neural voiceover with zero robotic artifacts.
3. **Distribution:** High-contrast thumbnails and burned kinetic captions.

## 🚀 3. Step-by-Step Implementation [2:30 - 5:00]
[Visual: Step-by-step split screen]
**Narrator:** Let's walk through the exact three steps you can replicate in under 10 minutes:
- Step A: Configure persistent local storage on Google Drive.
- Step B: Run 4-bit quantized LLMs (Qwen / Llama / Mistral) without needing $10,000 GPUs.
- Step C: Batch compile final high-definition MP4 renders.

## 🎯 4. Key Takeaways & Call to Action [5:00 - 6:00]
[Visual: End card with subscribe button and related video prompts]
**Narrator:** If this blueprint elevated your workflow, hit like, subscribe, and download the full Colab notebook below!
"""
        timestamp = time.strftime("%Y%m%d_%H%M%S")
        filename = f"{task_type.lower().replace(' ', '_')}_{timestamp}.md"
        filepath = os.path.join(self.output_dir, filename)
        
        with open(filepath, "w", encoding="utf-8") as f:
            f.write(script_body)
            
        print(f"✅ Output successfully saved to Google Drive: {filepath}")
        return script_body, filepath

writer_engine = AIWriterEngine()
```

```python
# Module 02 Gradio Interface
import gradio as gr

def run_writer(model_fam, task, topic, tone, words):
    text, path = writer_engine.generate_content(model_fam, task, topic, tone, int(words))
    return text, f"Saved to Drive: {path}"

with gr.Blocks(title="AI Creator OS — AI Writer", theme=gr.themes.Soft()) as demo:
    gr.Markdown("# ✍️ MODULE 02 — AI WRITER & LLM STUDIO")
    gr.Markdown("**Models & Tools:** Qwen 2.5, Llama 3.2, Gemma 2, Mistral 7B + Transformers, llama.cpp GGUF | *Free Colab Optimized*")
    
    with gr.Row():
        with gr.Column(scale=1):
            model_fam = gr.Dropdown(["Qwen 2.5 (7B/1.5B)", "Llama 3.2 (3B/1B)", "Gemma 2 (2B)", "Mistral 7B (Q4 GGUF)"], value="Qwen 2.5 (7B/1.5B)", label="Active LLM Architecture")
            task = gr.Dropdown(["YouTube Video Script (Timestamps + Visual Cues)", "Long-Form SEO Article (2000+ words)", "Viral Twitter/LinkedIn Thread", "Content Research & Summary", "Technical Documentation"], value="YouTube Video Script (Timestamps + Visual Cues)", label="Content Task")
            topic = gr.Textbox(label="Topic / Premise", value="How Autonomous AI Agents Are Transforming Content Creation in 2026")
            tone = gr.Dropdown(["Engaging & Energetic", "Cinematic & Storytelling", "Deep & Educational", "Executive & Authoritative"], value="Engaging & Energetic", label="Voice Tone")
            words = gr.Slider(200, 3000, value=800, step=50, label="Target Word Count")
            btn = gr.Button("🚀 Generate Full Content", variant="primary")
        with gr.Column(scale=2):
            out_md = gr.Markdown(label="Generated Output")
            out_file = gr.Textbox(label="Drive File Path")
            
    btn.click(run_writer, [model_fam, task, topic, tone, words], [out_md, out_file])

demo.launch(share=True, debug=False)
```
